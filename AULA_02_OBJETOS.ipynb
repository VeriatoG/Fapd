```python
import pandas as pd

datos_empleados = {
    "ID": [101, 102, 103, 104],
    "Nombre": ["Ana", "Luis", "Marta", "Carlos"],
    "Departamento": ["Ventas", "IT", "Ventas", "Marketing"],
    "Ciudad": ["Madrid", "Valencia", "Madrid", "Sevilla"],
    "Salario": [50000, 65000, 52000, 48000]
}

df_empleados = pd.DataFrame(datos_empleados)
```

```python
print(df_empleados.columns)

print(df_empleados["Salario"])
```

```python
print(df_empleados.iloc[0:2, 0:3])
```

```python
df_ventas = df_empleados[df_empleados["Departamento"] == "Ventas"]

df_ciudades = df_empleados[df_empleados["Ciudad"].isin(["Madrid", "Sevilla"])]
```

```python
datos_nuevo = {
    "ID": [105],
    "Nombre": ["Elena"],
    "Departamento": ["BI"],
    "Ciudad": ["Bilbao"],
    "Salario": [58000]
}

df_nuevo = pd.DataFrame(datos_nuevo)

df_empleados = pd.concat([df_empleados, df_nuevo], ignore_index=True)
```

```python
datos_proyectos = {
    "ID": [101, 103, 106],
    "Proyecto": ["IA Generativa", "ETL Cloud", "Data Lake"]
}

df_proyectos = pd.DataFrame(datos_proyectos)
```

```python
df_inner = pd.merge(df_empleados, df_proyectos, on="ID", how="inner")

print(df_inner)
```

```python
df_left = pd.merge(df_empleados, df_proyectos, on="ID", how="left")

print(df_left)
```

```python
def calcular_precio_final(precio, iva=0.21):
    """
    Calcula el precio final con IVA incluido.
    """
    return precio * (1 + iva)
```

```python
print(calcular_precio_final(100))

print(calcular_precio_final(100, 0.10))

print(calcular_precio_final(iva=0.04, precio=250))
```
